```python
import pandas as pd

#original dataset variable
orders_original = pd.read_csv('Files/fastdeliver_orders.csv')

#dublicated dataset variable
orders_dublicate = orders_original

#shows the first 5 rows of out dataset
orders_dublicate.head()
```

```python
#identify the number of dublicates in our dataset
orders_dublicate.duplicated().sum()
```

```python
#drop the duplicated rows in our dataset
orders_dublicate = orders_dublicate.drop_duplicates()
```

```python
#fill the null values in the rider_rating column with the median value of the column
orders_dublicate['rider_rating'] = orders_dublicate['rider_rating'].fillna(  
    orders_dublicate['rider_rating'].median()
 )
```

```python
#fill payment_method column null values with 'unknown'
orders_dublicate['payment_method'] = orders_dublicate['payment_method'].fillna('unknown')
```

```python
## removal of the ₦ sign and commas from the order_value column and convert it to float
orders_dublicate["order_value"] = (
    orders_dublicate["order_value"]
    .astype(str)
    .str.replace("₦", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
    .astype(float)
)
```

```python
#fill the null values in the order_value column with the median value of the column
orders_dublicate['order_value'] = orders_dublicate['order_value'].fillna(
    orders_dublicate['order_value'].median())
```

```python
#removal of the null values in the pickup_time column by converting it to datetime and filling the null values with the median date value of the column
orders_dublicate['pickup_time'] = pd.to_datetime(orders_dublicate['pickup_time'], errors='coerce', format='%H:%M')
```

```python
#fill the null values in the pickup_time column with the median value of the column
orders_dublicate["pickup_time"] = orders_dublicate["pickup_time"].fillna(
    orders_dublicate["pickup_time"].median()
)
```

```python
# Convert the pickup_time column to HH:MM format using strftime
orders_dublicate['pickup_time'] = orders_dublicate['pickup_time'].dt.strftime('%H:%M')
```

```python
#removal of the null values in the delivery_time column by converting it to datetime
orders_dublicate['delivery_time'] = pd.to_datetime(orders_dublicate['delivery_time'], errors='coerce', format='%H:%M')
```

```python
#fill the null values in the delivery_time column with the median value of the column
orders_dublicate["delivery_time"] = orders_dublicate["delivery_time"].fillna(
    orders_dublicate["delivery_time"].median()
)
```

```python
# Convert the delivery_time column to HH:MM format using strftime
orders_dublicate['delivery_time'] = orders_dublicate['delivery_time'].dt.strftime('%H:%M')
```

```python
# Convert the order_date column to datetime format
orders_dublicate['order_date'] = pd.to_datetime(orders_dublicate['order_date'], errors='coerce', format='%Y-%m-%d')
```

```python
# Convert the status column to lowercase and remove leading/trailing whitespace
orders_dublicate['status'] = orders_dublicate['status'].str.lower().str.strip()
```

```python
# Convert text columns to strings datatype
orders_dublicate = orders_dublicate.astype({
    'order_id': 'string',
    'customer_name': 'string',
    'pickup_zone': 'string',
    'delivery_zone': 'string',
    'rider_id': 'string',
    'status': 'string',
    'payment_method': 'string',
})
```
